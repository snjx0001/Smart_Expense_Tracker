# Smart Expense Tracker

Run the cells in order. The final cell starts the interactive menu.


```python
from expense_manager import ExpenseManager
from budget_manager import BudgetManager
from reports import ReportGenerator
from storage import Storage
from dashboard import Dashboard
from validation import (
    validate_amount,
    validate_category,
    validate_date,
    validate_description,
)

manager = ExpenseManager()
budget = BudgetManager()
reports = ReportGenerator()
storage = Storage()
dashboard = Dashboard()

manager.expenses = storage.load()

print("======================================")
print("       SMART EXPENSE TRACKER")
print("======================================")
print("System loaded successfully!")
print("Expenses loaded:", len(manager.expenses))
```

```python
def view_expenses(expenses=None):
    records = manager.expenses if expenses is None else expenses
    print("\n========== ALL EXPENSES ==========")

    if not records:
        print("No expenses found.")
        return

    for i, expense in enumerate(records, start=1):
        print(f"\nExpense {i}")
        print(f"Amount      : ₹{expense['amount']:.2f}")
        print(f"Category    : {expense['category']}")
        print(f"Description : {expense['description']}")
        print(f"Date        : {expense['date']}")
```

```python
def add_expense_jupyter():
    print("\n========== ADD EXPENSE ==========")

    try:
        amount = validate_amount(input("Enter amount: "))
        category = validate_category(input("Enter category: "))
        description = validate_description(input("Enter description: "))
        date = validate_date(input("Enter date (YYYY-MM-DD): "))

        manager.add_expense(amount, category, description, date)
        storage.save(manager.expenses)
        print("\nExpense added and saved successfully!")

    except ValueError as error:
        print("\nError:", error)
```

```python
def update_expense_jupyter():
    print("\n========== UPDATE EXPENSE ==========")

    if not manager.expenses:
        print("No expenses available.")
        return

    view_expenses()

    try:
        index = int(input("\nEnter expense number to update: ")) - 1
        amount = validate_amount(input("New amount: "))
        category = validate_category(input("New category: "))
        description = validate_description(input("New description: "))
        date = validate_date(input("New date (YYYY-MM-DD): "))

        manager.update_expense(index, amount, category, description, date)
        storage.save(manager.expenses)
        print("\nExpense updated and saved successfully!")

    except (ValueError, IndexError) as error:
        print("\nError:", error)
```

```python
def delete_expense_jupyter():
    print("\n========== DELETE EXPENSE ==========")

    if not manager.expenses:
        print("No expenses available.")
        return

    view_expenses()

    try:
        index = int(input("\nEnter expense number to delete: ")) - 1
        manager.delete_expense(index)
        storage.save(manager.expenses)
        print("\nExpense deleted and saved successfully!")

    except (ValueError, IndexError) as error:
        print("\nError:", error)
```

```python
def search_expense_jupyter():
    print("\n========== SEARCH EXPENSE ==========")

    keyword = input("Enter category or description to search: ")
    results = manager.search(keyword)

    if not results:
        print("No matching expenses found.")
        return

    view_expenses(results)
```

```python
def show_summary():
    dashboard.display(manager.expenses, budget, reports)
```

```python
def set_budget_jupyter():
    print("\n========== SET BUDGET ==========")

    try:
        amount = validate_amount(input("Enter monthly budget: "))
        budget.set_budget(amount)
        print(f"Monthly budget set to ₹{amount:.2f}")

    except ValueError as error:
        print("Error:", error)


def show_budget():
    print("\n========== BUDGET STATUS ==========")

    total = manager.total_expenses()
    remaining = budget.remaining_budget(total)

    print(f"Budget          : ₹{budget.budget:.2f}")
    print(f"Total Expenses  : ₹{total:.2f}")
    print(f"Remaining       : ₹{remaining:.2f}")

    if budget.budget == 0:
        print("No budget set.")
    elif remaining < 0:
        print("Budget exceeded!")
    else:
        print("You are within your budget.")
```

```python
def save_expenses():
    storage.save(manager.expenses)
    print("\nAll expenses saved successfully!")
```

```python
def run_expense_tracker():
    while True:
        print("\n======================================")
        print("       SMART EXPENSE TRACKER")
        print("======================================")
        print("1. Add Expense")
        print("2. View Expenses")
        print("3. Update Expense")
        print("4. Delete Expense")
        print("5. Search Expense")
        print("6. Expense Summary")
        print("7. Set Budget")
        print("8. Budget Status")
        print("9. Save Expenses")
        print("10. Exit")
        print("======================================")

        choice = input("Enter your choice: ").strip()

        if choice == "1":
            add_expense_jupyter()
        elif choice == "2":
            view_expenses()
        elif choice == "3":
            update_expense_jupyter()
        elif choice == "4":
            delete_expense_jupyter()
        elif choice == "5":
            search_expense_jupyter()
        elif choice == "6":
            show_summary()
        elif choice == "7":
            set_budget_jupyter()
        elif choice == "8":
            show_budget()
        elif choice == "9":
            save_expenses()
        elif choice == "10":
            save_expenses()
            print("\nThank you for using Smart Expense Tracker!")
            break
        else:
            print("\nInvalid choice. Please enter a number from 1 to 10.")
```

```python
run_expense_tracker()
```
